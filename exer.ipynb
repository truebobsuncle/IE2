```python
import numpy as np
arr = np.random.randint(1, 101, 20)
print("Исх массив:")
print(arr)

a = arr % 5 == 0
print("\nЭл, дел на 5:")
print(arr[a])

arr[a] = -1
print("\nОбнов.массив (эл, дел на 5, зам на -1):")
print(arr)
```

```python
import numpy as np
arr1 = np.random.randint(0, 51, 5)
arr2 = np.random.randint(0, 51, 5)
print(arr1)
print(arr2)

a = np.vstack([arr1, arr2])
print("\nДвумерный массив:")
print(a)

b, c = a[0, :], a[1, : ]
print("\nРазделенные массивы:")
print("Первый:", b)
print("Второй:", c)
```

```python
import numpy as np
arr = np.linspace(-10, 10, 50)
print("Массив из 50 чисел от -10 до 10:")
print(arr)

total_sum = np.sum(arr)
print(f"\nСумма всех элементов: {total_sum}")
plus_sum = np.sum(arr[arr > 0])
print(f"Сумма положительных элементов: {plus_sum:.4f}")
minus_sum = np.sum(arr[arr < 0])
print(f"Сумма отрицательных элементов: {minus_sum:.4f}")
```

```python
import numpy as np
edmat = np.eye(4)
print("Единичная матрица 4x4:")
print(edmat)

d = np.diag([5, 10, 15, 20])
print("\nДиагональная матрица:")
print(d)

sum_edmat = np.sum(edmat)
sum_d = np.sum(d)

print(f"\nСумма элементов единичной матрицы: {sum_edmat}")
print(f"Сумма элементов диагональной матрицы: {sum_d}")
print(f"Разница: {sum_d - sum_edmat}")
```

```python
import numpy as np
A = np.random.randint(1, 21, (3, 3))
B = np.random.randint(1, 21, (3, 3))

print("Матрица А:")
print(A)
print("\nМатрица В:")
print(B)

smatr = A + B
print("\nСумма А + В:")
print(smatr)

raznat = A - B
print("\nРазность А - В:")
print(raznat)

poelement = A * B
print("\nПоэлементное произведение:")
print(poelement)
```

```python
import numpy as np
A = np.random.randint(1, 11, (2, 3))
print("Матрица А (2x3):")
print(A)

B = np.random.randint(1, 11, (3, 2))
print("\nМатрица В (3x2):")
print(B)

result = np.dot(A, B)
print("\nРезультат умножения А * В (2x2):")
print(result)
```

```python
import numpy as np
A = np.random.rand(3, 3)
print("Матрица А:")
print(A)

opr = np.linalg.det(A)
print(f"\nОпределитель матрицы: {opr:.4f}")

if opr == 0:
    print("\nМатрица вырождена")
else:
    opr = np.linalg.inv(A)
    print("\nОбратная матрица:\n", opr)
```

```python
import numpy as np
A = np.random.randint(1, 51, (4, 4))
print("Матрица 4x4:")
print(A)

transposed = np.transpose(A)
print("\nТранспонированная матрица:")
print(transposed)

trace = np.trace(A)
print(f"\nСлед матрицы: {trace}")
```

```python
import numpy as np


A = np.array([[2, 3, -1], [4, -1, 2], [-3, 5, 4]])
B = np.array([[5], [6], [-2]])

print("Матрица коэффициентов А:")
print(A)
print("\nВектор правой части В:")
print(B)


res = np.linalg.solve(A, B)

print("\nРешение системы:")
print(f"x = {res[0][0]:.4f}")
print(f"y = {res[1][0]:.4f}")
print(f"z = {res[2][0]:.4f}")
```
